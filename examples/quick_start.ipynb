# DandeLiion API - Quick Start Example

```python
import os

import dandeliion.client as dandeliion
```

### Authentication
Define the simulator object by providing the API URL and API token:

```python
api_url = "https://api.dandeliion.com"
api_key = os.environ["DANDELIION_API_KEY"]
simulator = dandeliion.Simulator(api_url, api_key)
```

### Start the simulation in the cloud

```python
# 1C discharge for 1 hour or until the `Lower voltage cut-off [V]` from BPX file will be used by default
solution = dandeliion.solve(
    simulator=simulator,
    params="BPX/lfp_18650_cell_BPX.json",
)

print("Status:", solution.status)
```

### Results

```python
# Print the final values of time, voltage, and temperature
print(f"Final time [s]:        {solution['Time [s]'][-1]:.3f}")
print(f"Final voltage [V]:     {solution['Voltage [V]'][-1]:.3f}")
print(f"Final temperature [K]: {solution['Temperature [K]'][-1]:.3f}")
```
